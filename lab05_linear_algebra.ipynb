```python
import numpy as np

# Coefficient matrix A
A = np.array([[3, 2], 
              [1, 4]])

# Constants vector b
b = np.array([12, 10])

# Solve
solution = np.linalg.solve(A, b)
print(f"x = {solution[0]}, y = {solution[1]}")
```

```python
A3 = np.array([[2, 1, -1], 
               [-3, -1, 2], 
               [-2, 1, 2]])
b3 = np.array([8, -11, -3])

solution_3 = np.linalg.solve(A3, b3)
print("Solution for 3x3 system:", solution_3)
```

np.linalg.solve() is generally preferred because it is computationally faster and more numerically stable. Calculating the inverse explicitly (np.linalg.inv(A) @ b) involves calculating the full matrix inverse, which takes roughly twice as many operations and introduces more floating-point rounding errors, especially for large or ill-conditioned matrices. solve() uses optimized algorithms like LU decomposition to directly find x without inverting A.

(i) Use a regression design matrix to estimate coefficients.

```python
X_data = np.array([30, 45, 60, 100, 150, 200, 240, 250, 300, 320])
Y_data = np.array([25, 20, 40, 65, 120, 190, 200, 200, 260, 220])

# Create design matrix (add column of 1s for intercept)
X_design = np.column_stack((np.ones(len(X_data)), X_data))

# Estimate coefficients: Beta = (X^T * X)^-1 * X^T * Y
# (Using the @ operator for matrix multiplication)
beta = np.linalg.inv(X_design.T @ X_design) @ X_design.T @ Y_data

print(f"Intercept (Beta 0): {beta[0]:.4f}")
print(f"Slope (Beta 1): {beta[1]:.4f}")
# The model is: Consumption = Beta0 + Beta1 * Income
```

(ii) Estimate consumption for income = 181.

```python
income_new = 181
predicted_consumption = beta[0] + beta[1] * income_new
print(f"Predicted consumption for income 181: {predicted_consumption:.2f}")
```

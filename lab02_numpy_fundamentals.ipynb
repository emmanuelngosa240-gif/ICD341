```python
# Define the array first
marks = np.array([45, 67, 82, 56, 91, 73, 64, 88])

# 1. Display the first mark
print("First mark:", marks[0])

# 2. Display the last mark
print("Last mark:", marks[-1])

# 3. Display the first three marks
print("First three:", marks[:3])

# 4. Display marks from position 3 to 6 (indices 2 to 5)
print("Positions 3 to 6:", marks[2:6])

# 5. Display every second mark
print("Every second mark:", marks[::2])
```

```python
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Std Dev:", np.std(marks))
print("Min:", np.min(marks))
print("Max:", np.max(marks))
print("Sum:", np.sum(marks))
```

• Mean (70.75): The average mark of the students.
• Median (71.5): The middle value; since it's very close to the mean, the data is fairly symmetric.
• Standard Deviation (14.88): The typical distance of a mark from the mean. Most students scored within 14.88 marks of 70.75.
• Min (45) & Max (91): The range of the scores. The lowest was 45, the highest was 91.
• Sum (566): The total of all marks combined.

```python
data = np.array([
    [45, 67, 82],
    [56, 91, 73],
    [64, 88, 76]
])

print("Shape:", data.shape)
print("Dimensions:", data.ndim)
```

```python
# 1. The first row
print("First row:", data[0])

# 2. The second column (all rows, column index 1)
print("Second column:", data[:, 1])

# 3. The value 91 (row index 1, column index 1)
print("The value 91:", data[1, 1])

# 4. The first two rows
print("First two rows:\n", data[:2])

# 5. The last two columns
print("Last two columns:\n", data[:, -2:])
```

```python
# Mean for each subject (column)
print("Subject means (axis=0):", np.mean(data, axis=0))

# Mean for each student (row)
print("Student means (axis=1):", np.mean(data, axis=1))
```

Axis 0 vs Axis 1:
• axis=0 collapses the rows. It calculates the mean down each column, giving one value per subject.
• axis=1 collapses the columns. It calculates the mean across each row, giving one value per student.

```python
import matplotlib.pyplot as plt

# Generate 1,000 observations from a normal distribution
x = np.random.normal(loc=50, scale=10, size=1000)

# Calculate statistics
print("Mean:", np.mean(x))
print("Median:", np.median(x))
print("Std Dev:", np.std(x))
print("Min:", np.min(x))
print("Max:", np.max(x))

# Create the histogram
plt.hist(x, bins=30, edgecolor='black')
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Distribution of Simulated Data")
plt.show()
```

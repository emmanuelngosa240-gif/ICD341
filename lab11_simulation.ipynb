```python
import numpy as np

np.random.seed(42)

print("--- 1. Inverse Transform for Exponential (rate 3) ---")
# U ~ Uniform(0,1), X = -ln(U)/rate
u = np.random.uniform(0, 1, 5000)
exp_samples = -np.log(u) / 3
print(f"Mean of generated samples: {np.mean(exp_samples):.4f} (Target is 1/3 = 0.3333)\n")

print("--- 2. Categorical Distribution [0.1, 0.6, 0.3] ---")
probs = [0.1, 0.6, 0.3]
samples = np.random.choice([0, 1, 2], size=10000, p=probs)
print(f"Sample proportions: {np.bincount(samples) / 10000}\n")

print("--- 3. Rejection Sampling for Beta(2,2) ---")
# Beta(2,2) pdf is proportional to x(1-x). Max value is 1.5 at x=0.5
def beta_22_pdf(x): return 6 * x * (1 - x)
samples = []
while len(samples) < 5000:
    # Proposal from Uniform(0,1) and Uniform(0, 1.5)
    x_prop = np.random.uniform(0, 1)
    y_prop = np.random.uniform(0, 1.5) 
    if y_prop < beta_22_pdf(x_prop):
        samples.append(x_prop)
print(f"Mean of rejection samples: {np.mean(samples):.4f} (Target is 0.5)\n")
```

4. Four main particle-filter steps:
   - Initialization: Draw initial particles from the prior distribution.
   - Prediction (Propagate): Move the particles forward using the state transition model.
   - Update (Weighting): Calculate the likelihood of the observed data given each particle's state, and assign weights.
   - Resampling: Eliminate particles with low weights and duplicate particles with high weights to prevent particle degeneracy. 
5. What a non-uniform SBC rank histogram indicates:
A non-uniform shape indicates that the Bayesian model or the sampler is biased (the posterior is not well-calibrated). A U-shape means the posterior is too narrow (overconfident), while a bell shape means it is too wide (underconfident).

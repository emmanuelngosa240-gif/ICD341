```python
import numpy as np

# Set a seed for reproducibility
np.random.seed(42)

print("--- 1. 10,000 observations from N(50, 10^2) ---")
obs = np.random.normal(loc=50, scale=10, size=10000)
print(f"Sample Mean: {np.mean(obs):.2f}")
print(f"Sample Std Dev: {np.std(obs):.2f}\n")

print("--- 2. 10,000 Bernoulli observations with p=0.3 ---")
bernoulli_obs = np.random.binomial(n=1, p=0.3, size=10000)
prop_success = np.mean(bernoulli_obs)
print(f"Sample proportion of successes: {prop_success:.4f}\n")

print("--- 3. Monte Carlo estimation of integral of x^3 from 0 to 1 ---")
# Generate random x values between 0 and 1
x_vals = np.random.uniform(0, 1, 100000)
# Evaluate x^3 and take the mean, then multiply by the interval length (1 - 0)
integral_estimate = np.mean(x_vals**3) * 1
print(f"Estimated Integral: {integral_estimate:.4f} (Exact is 0.25)\n")

print("--- 4. Estimate Pi using 1 million points ---")
n_points = 1_000_000
# Generate random points in a 2x2 square from -1 to 1
x_pts = np.random.uniform(-1, 1, n_points)
y_pts = np.random.uniform(-1, 1, n_points)
# Check if points fall inside the unit circle (radius 1)
inside = (x_pts**2 + y_pts**2) <= 1
# Ratio of inside points to total points is Area(circle)/Area(square) = Pi / 4
pi_estimate = 4 * np.mean(inside)
print(f"Estimated Pi: {pi_estimate:.5f}\n")
```

```python
print("--- 5. Estimate P(X > 1.96) where X ~ N(0,1) ---")
# Generate 1,000,000 standard normal observations
z = np.random.normal(0, 1, 1000000)
prob_est = np.mean(z > 1.96)
print(f"Estimated Probability: {prob_est:.4f} (Exact is ~0.0250)\n")

print("--- 6. Importance Sampling for E(X^2) where X ~ U(0,1) ---")
# Target: E[X^2] under U(0,1). 
# Instead of sampling from U(0,1), let's use a proposal distribution q(x) = 2x (Beta(2,1))
# This puts more samples at higher x values where x^2 is larger.
n_samples = 100000
# Sample from q(x)
x_imp = np.random.beta(2, 1, n_samples)
# Importance weights: target density f(x)=1, proposal density q(x)=2x
weights = 1 / (2 * x_imp)
# Calculate estimate
is_estimate = np.mean(x_imp**2 * weights)
print(f"Importance Sampling Estimate: {is_estimate:.4f} (Exact is 1/3 = 0.3333)\n")
```

Difference between Importance Sampling and Control Variates:
• Importance Sampling is a variance reduction technique where you sample from a different probability distribution (the proposal distribution) that focuses on the "important" regions of the target function. You then weight the samples to correct for the fact that you didn't sample from the original distribution. It is highly effective for rare-event simulation.
• Control Variates involves finding a secondary variable (the control variate) that is highly correlated with your variable of interest but whose expected value is exactly known. By subtracting the control variate and adding its known expectation, you mathematically cancel out a large portion of the variance in your estimate, leading to higher precision without changing the sampling distribution.

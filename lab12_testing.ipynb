```python
import numpy as np
import pytest

print("--- 1 & 2. Testing functions ---")
def calculate_median(data):
    if len(data) == 0:
        raise ValueError("Cannot calculate median of empty array")
    return np.median(data)

def estimate_probability(successes, trials):
    if trials == 0:
        return 0
    return successes / trials

# Here are the tests (normally run via terminal with pytest)
def test_median():
    assert calculate_median([1, 2, 3, 4, 5]) == 3
    with pytest.raises(ValueError):
        calculate_median([])

def test_probability():
    p = estimate_probability(3, 10)
    assert 0 <= p <= 1

# Run the tests manually in the notebook
test_median()
test_probability()
print("All tests passed successfully!\n")

print("--- 3. Transformer for Standard Deviation ---")
class StdDevScaler:
    def fit(self, X):
        self.scales_ = np.std(X, axis=0)
        return self
    
    def transform(self, X):
        scales = np.where(self.scales_ == 0, 1.0, self.scales_)
        return X / scales

data = np.array([[10, 100], [20, 200], [30, 300]])
scaler = StdDevScaler().fit(data)
print("Scaled Data:\n", scaler.transform(data))
```

pyproject.toml and editable installation:
Editable installation (pip install -e .) allows you to work on a package while it's still installed in your environment. Instead of copying files to site-packages, pip creates a link to your working directory, so any changes you make to the code take effect immediately without needing to reinstall.
 
A minimal pyproject.toml looks like this:
[build-system]
requires = ["setuptools>=61.0"]
build-backend = "setuptools.build_meta"

[project]
name = "my_icd341_package"
version = "0.1.0"

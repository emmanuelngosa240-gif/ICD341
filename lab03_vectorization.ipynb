```python
import numpy as np

x = np.array([1, 2, 3, 4, 5])
result = x ** 2
print("Squared:", result)

# Task 1: Calculate y = 2x + 5
y = 2 * x + 5
print("2x + 5:", y)
```

```python
marks = np.array([40, 55, 62, 70, 81, 90])

# Add 5 marks to everyone
adjusted = marks + 5
print("Adjusted marks:", adjusted)

# Calculate percentage
percentage = marks / 100
print("Percentages:", percentage)

# Task 2: Standardize the marks
z = (marks - np.mean(marks)) / np.std(marks)
print("Standardized marks (z):", z)

# Verify the mean of z is approximately zero
print("Mean of z:", np.mean(z))
```

```python
data = np.array([
    [60, 70, 80],
    [50, 65, 75],
    [90, 85, 95]
])

adjustment = np.array([5, 0, 2])

result = data + adjustment
print("Original data:\n", data)
print("\nAdjustment vector:", adjustment)
print("\nResulting data:\n", result)
```

```python
data2 = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

means = data2.mean(axis=0)
print("Column means:", means)

centered = data2 - means
print("\nCentered data:\n", centered)

# Task 3: Verify the mean of each column is now approximately zero
print("\nMeans of centered data:", centered.mean(axis=0))
```

```python
x = np.arange(1_000_000)

# Loop version
result_loop = []
for value in x:
    result_loop.append(value**2 + 2*value + 1)

# Vectorized version
result_vectorized = x**2 + 2*x + 1

# Check if they agree
print("Are they equal?", np.allclose(result_loop, result_vectorized))
```

```python
np.random.seed(123)
x_large = np.random.normal(100, 15, 100_000)

# Calculate standardized values
z = (x_large - x_large.mean()) / x_large.std()

print("Mean of z:", z.mean())
print("Std Dev of z:", z.std())
```

```python
# 1. The basic function
def standardize(x):
    return (x - np.mean(x)) / np.std(x)

# 2. Test it using three different arrays
arr1 = np.array([10, 20, 30, 40, 50])
arr2 = np.array([1, 2, 3])
arr3 = np.array([100, 200, 300, 400])

print("Test 1:", standardize(arr1))
print("Test 2:", standardize(arr2))
print("Test 3:", standardize(arr3))

print("\n--- Modified for 2D Data ---")

# 3. Modified function to standardize column by column
def standardize_columns(x):
    # axis=0 calculates mean/std down each column
    # keepdims=True ensures the shape stays (1, n_columns) for easy broadcasting
    mean = np.mean(x, axis=0, keepdims=True)
    std = np.std(x, axis=0, keepdims=True)
    return (x - mean) / std

# Test the modified function on a 2D dataset
data_2d = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

standardized_2d = standardize_columns(data_2d)
print("2D Data:\n", data_2d)
print("\nStandardized 2D Data:\n", standardized_2d)

# Verify the columns have mean 0 and std 1
print("\nMean of each column:", np.mean(standardized_2d, axis=0))
print("Std Dev of each column:", np.std(standardized_2d, axis=0))
```

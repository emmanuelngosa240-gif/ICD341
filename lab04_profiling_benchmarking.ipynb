```python
import numpy as np

x = np.arange(1_000_000)

# Benchmark the operation
%timeit x ** 2
```

```python
# The loop function
def square_loop(x):
    result = []
    for value in x:
        result.append(value ** 2)
    return result

# The NumPy function
def square_numpy(x):
    return x ** 2

# Benchmark both
print("Python Loop:")
%timeit square_loop(x)

print("\nNumPy Vectorization:")
%timeit square_numpy(x)
```

```python
x1 = np.arange(100)
x2 = np.arange(10_000)
x3 = np.arange(1_000_000)

print("Array size 100:")
%timeit x1 ** 2

print("\nArray size 10,000:")
%timeit x2 ** 2

print("\nArray size 1,000,000:")
%timeit x3 ** 2
```

```python
data = np.random.normal(size=1_000_000)

print("np.mean:")
%timeit np.mean(data)

print("\nnp.std:")
%timeit np.std(data)

print("\nnp.sum:")
%timeit np.sum(data)
```

```python
# Define the function right in the notebook
import numpy as np
def calculate_statistics(x):
    total = np.sum(x)
    mean = np.mean(x)
    variance = np.var(x)
    result = np.sqrt(variance)
    return total, mean, result

# Generate the data
data = np.random.normal(size=1_000_000)

# Run the profiler
%lprun -f calculate_statistics calculate_statistics(data)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

data = np.array([3, 4, 5, 7, 8, 9])
n_bootstraps = 10000
np.random.seed(42)

print("--- 1. Bootstrap the mean ---")
# Resample with replacement
boot_means = [np.mean(np.random.choice(data, size=len(data), replace=True)) for _ in range(n_bootstraps)]
# 95% Percentile interval
ci_mean = np.percentile(boot_means, [2.5, 97.5])
print(f"95% CI for the Mean: {ci_mean}")

print("\n--- 2. Bootstrap the median ---")
boot_medians = [np.median(np.random.choice(data, size=len(data), replace=True)) for _ in range(n_bootstraps)]
ci_median = np.percentile(boot_medians, [2.5, 97.5])
print(f"95% CI for the Median: {ci_median}")

print("\n--- 3. Parametric bootstrap from a fitted normal model ---")
# Fit normal model to data
mu_hat, std_hat = np.mean(data), np.std(data, ddof=1)
# Generate parametric bootstrap samples
param_boot_means = [np.mean(np.random.normal(mu_hat, std_hat, size=len(data))) for _ in range(n_bootstraps)]
ci_param = np.percentile(param_boot_means, [2.5, 97.5])
print(f"95% CI for Parametric Mean: {ci_param}")
```

Data leakage occurs if you perform feature scaling (like standardization) or imputation on the entire dataset before splitting it into train and test sets. The test set's mean and variance "leak" into the training process. To prevent this, you must fit the scaler only on the training data and apply that transformation to the test data.

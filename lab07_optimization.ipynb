```python
import numpy as np
from scipy.optimize import minimize
from scipy.stats import norm

print("--- 1. Gradient Descent for f(x) = (x - 10)^2 ---")
# Derivative is f'(x) = 2*(x - 10)
def gradient_descent(start_x, learning_rate, epochs):
    x = start_x
    for _ in range(epochs):
        grad = 2 * (x - 10)
        x = x - learning_rate * grad
    return x

min_x = gradient_descent(start_x=0, learning_rate=0.1, epochs=100)
print(f"Minimum found at x = {min_x:.4f} (Should be close to 10)\n")


print("--- 2. Compare BFGS and CG for f(x,y) = (x-4)^2 + (y-6)^2 ---")
def f(params):
    x, y = params
    return (x - 4)**2 + (y - 6)**2

res_bfgs = minimize(f, x0=[0, 0], method='BFGS')
print(f"BFGS Result: x={res_bfgs.x[0]:.2f}, y={res_bfgs.x[1]:.2f} (Iterations: {res_bfgs.nit})")

res_cg = minimize(f, x0=[0, 0], method='CG')
print(f"CG Result:   x={res_cg.x[0]:.2f}, y={res_cg.x[1]:.2f} (Iterations: {res_cg.nit})\n")


print("--- 4. MLE for the mean of normal observations ---")
# Generate fake data with a true mean of 15
np.random.seed(42)
data = np.random.normal(loc=15, scale=5, size=100)

# We want to minimize the Negative Log-Likelihood (NLL)
def neg_log_likelihood(mu):
    # Assuming we know the standard deviation is 5
    return -np.sum(norm.logpdf(data, loc=mu, scale=5))

result = minimize(neg_log_likelihood, x0=[0])
print(f"MLE Estimated Mean: {result.x[0]:.4f} (True mean was 15)\n")
```

```python
print("--- 5. Sigmoid Function ---")
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

print("Sigmoid(0):", sigmoid(0))
print("Sigmoid([2, -2]):", sigmoid(np.array([2, -2])))
print("\n")


print("--- 6. Logistic Regression using MLE ---")
X = np.array([1, 2, 3, 4, 5, 6, 7, 8])
Y = np.array([0, 0, 0, 0, 1, 1, 1, 1])

# Add a column of 1s for the intercept (Beta 0)
X_design = np.column_stack((np.ones(len(X)), X))

# Negative Log-Likelihood function
def nll_logistic(beta):
    p = sigmoid(X_design @ beta)
    # Clip probabilities to avoid log(0) which is negative infinity
    epsilon = 1e-15 
    p = np.clip(p, epsilon, 1 - epsilon)
    return -np.sum(Y * np.log(p) + (1 - Y) * np.log(1 - p))

# Initial guess for beta: [intercept, slope]
initial_beta = np.zeros(2)

# Minimize the negative log-likelihood
result = minimize(nll_logistic, initial_beta)
print(f"Intercept (Beta0): {result.x[0]:.4f}")
print(f"Slope (Beta1): {result.x[1]:.4f}")
```
